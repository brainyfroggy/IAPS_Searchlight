```python
import numpy as np
import matplotlib.pyplot as plt

# Generate example beta values (mostly positive, a few negative)
np.random.seed(11)
n_voxels = 15
betas = np.random.normal(0.6, 0.6, n_voxels)  # mostly positive

# Color: dark blue for positive, dark orange for negative
colors = ['#2B5C8A' if b >= 0 else '#D2691E' for b in betas]

fig, ax = plt.subplots(figsize=(4, 3.5))
ax.bar(range(1, n_voxels + 1), betas, color=colors, edgecolor='none', width=0.8)
ax.axhline(0, color='black', linewidth=0.5)

# X-axis: show 1, ..., 10, ...
ax.set_xticks([1, 5, 10, 15])
ax.set_xticklabels(['1', '...', '10', '...'])
ax.set_xlabel('Voxel Index', fontsize=18)
ax.set_ylabel('Beta Value', fontsize=18)

ax.tick_params(axis='both', labelsize=16)

ax.set_ylim(-1.5, 3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.savefig('beta_bar.png', dpi=200)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# Publish-ready EEG-style stacked trace figure
# Randomized traces each run, but always with a clear effect
# around 450–550 ms
# ============================================================

def smooth_signal(x, sigma=4):
    radius = int(4 * sigma)
    kx = np.arange(-radius, radius + 1)
    kernel = np.exp(-(kx**2) / (2 * sigma**2))
    kernel /= kernel.sum()
    return np.convolve(x, kernel, mode="same")

# ---------- time axis ----------
t = np.linspace(-300, 2000, 575)
n = len(t)

# ---------- channels shown ----------
labels = ["Fp1", "F7", "FC5"]
ypos   = [3.0, 2.0, 0.0]

# ---------- random generator ----------
rng = np.random.default_rng()

# ---------- make random EEG-like curves ----------
curves = []
for i in range(3):
    base = rng.normal(0, 1, n)
    base = smooth_signal(base, sigma=5)
    base = base / np.std(base) * 0.09

    drift = 0.03 * np.sin(2 * np.pi * (t / 700 + rng.uniform(0, 1)))
    noise = smooth_signal(rng.normal(0, 1, n), sigma=2) * 0.018

    # force a visible positive or negative ERP-like peak around 500 ms
    peak_center = rng.uniform(470, 530)
    peak_width  = rng.uniform(30, 55)
    peak_sign   = rng.choice([-1, 1])
    peak_amp    = rng.uniform(0.22, 0.34)
    peak = peak_sign * peak_amp * np.exp(-(t - peak_center)**2 / (2 * peak_width**2))

    # optional smaller opposite rebound nearby for more natural shape
    rebound_center = peak_center + rng.uniform(35, 75)
    rebound_width  = rng.uniform(25, 45)
    rebound_amp    = -peak_sign * rng.uniform(0.05, 0.12)
    rebound = rebound_amp * np.exp(-(t - rebound_center)**2 / (2 * rebound_width**2))

    y = ypos[i] + base + drift + noise + peak + rebound
    curves.append(y)

# ---------- figure ----------
fig, ax = plt.subplots(figsize=(6.4, 6.4), dpi=200)
fig.patch.set_facecolor("white")
ax.set_facecolor("white")

# ---------- vertical grid lines ----------
grid_x = [-300, 0, 600, 1700, 2000]
for gx in grid_x:
    ax.axvline(gx, color="0.75", lw=1.0, linestyle=(0, (1.2, 6)), zorder=0)

# ---------- horizontal dotted separators ----------
for gy in [2.5, 1.5, 0.5]:
    ax.axhline(gy, color="0.75", lw=1.0, linestyle=(0, (1.2, 6)), zorder=0)

# ---------- dashed line at 0 ms ----------
ax.axvline(0, color="0.2", lw=1.6, linestyle="--", zorder=1)

# ---------- purple highlight window around 500 ms ----------
ax.axvspan(400, 600, color="#d8c4e6", alpha=0.75, zorder=0.5)

# ---------- traces ----------
for y in curves:
    ax.plot(t, y, color="black", lw=2.2, solid_capstyle="round", zorder=3)

# ---------- axes styling ----------
ax.set_xlim(-300, 2000)
ax.set_ylim(-0.7, 3.55)

ax.set_xticks([0, 500, 2000])
ax.set_xticklabels(["0ms", "500ms", "2000ms"], fontsize=18)

ax.set_yticks([])

# ---------- left labels ----------
ax.text(-340, ypos[0], "Fp1", ha="right", va="center", fontsize=26)
ax.text(-340, ypos[1], "F7",  ha="right", va="center", fontsize=26)
ax.text(-340, ypos[2], "FC5", ha="right", va="center", fontsize=26)

# ---------- ellipsis ----------
ax.text(-525, 1.03, "⋮", fontsize=34, va="center", ha="center")
ax.text(980,  1.03, "⋮", fontsize=34, va="center", ha="center")

# ---------- x label ----------
ax.set_xlabel("Time (-300ms to 2000ms)", fontsize=20)

# ---------- spines ----------
for spine in ["top", "right"]:
    ax.spines[spine].set_visible(False)

ax.spines["left"].set_linewidth(2.0)
ax.spines["bottom"].set_linewidth(2.0)

ax.tick_params(axis="x", width=1.5, length=0, pad=8)
ax.tick_params(axis="y", length=0)

plt.subplots_adjust(left=0.28, bottom=0.18, right=0.96, top=0.92)
plt.show()
```
